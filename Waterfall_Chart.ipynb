```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

# Datos
data = pd.DataFrame({
    'Concepto': [
        'Ingresos totales',
        'Coste de ventas',
        'Gastos operativos',
        'Otros ingresos',
        'Impuestos',
        'Beneficio neto'
    ],
    'Valor': [189395, -100810, -30516, -0.45, -9711, 48214]
})

# Medidas: todo relativo salvo el último (total)
measures = ['relative'] * (len(data) - 1) + ['total']

# Cálculos para hover
vals = data['Valor'].to_numpy(dtype=float)
cum = np.cumsum(vals)
base_ingresos = abs(vals[0]) if vals.size else 1.0
share = (vals / base_ingresos) * 100.0

# Formato europeo
def fmt(x):
    return f"{x:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")

# Colores
color_inc = "#2ecc71"
color_dec = "#e74c3c"
color_tot = "#34495e"

fig = go.Figure(go.Waterfall(
    name="Estado de resultados",
    orientation="v",
    measure=measures,
    x=data['Concepto'],
    y=data['Valor'],
    text=[fmt(v) for v in vals],
    textposition="outside",
    textfont=dict(size=16),
    connector={"line": {"width": 1, "color": "rgba(120,120,120,0.6)"}},
    increasing={"marker": {"color": color_inc, "line": {"width": 0}}},
    decreasing={"marker": {"color": color_dec, "line": {"width": 0}}},
    totals={"marker": {"color": color_tot, "line": {"width": 0}}},
    customdata=np.column_stack([vals, cum, share]),
    hovertemplate=(
        "<b>%{x}</b><br>"
        "Movimiento: %{customdata[0]:,.2f} €<br>"
        "Acumulado: %{customdata[1]:,.2f} €<br>"
        "Sobre ingresos: %{customdata[2]:.2f}%"
        "<extra></extra>"
    )
))

# Diseño
fig.update_layout(
    template="plotly_white",
    title="<b>Waterfall Chart de resultados empresariales de Apple de los dos últimos semestres (Q1 2025 y Q2 2025) </b>",
    height=540,
    margin=dict(l=80, r=40, t=70, b=60),
    font=dict(family="Inter, Segoe UI, Roboto, Arial", size=13),
    xaxis=dict(
        title=None,
        tickangle=-15,
        categoryorder="array",
        categoryarray=list(data['Concepto']),
        showline=True,
        linecolor="rgba(0,0,0,0.3)",
        linewidth=0.1
    ),
    yaxis=dict(
        title="Montos en millones de euros",
        zeroline=True,
        zerolinewidth=2,
        zerolinecolor="black",
        gridcolor="rgba(0,0,0,0.06)",
        tickformat=",.0f"
    ),
    showlegend=False,
    plot_bgcolor="white",
)

fig.update_traces(cliponaxis=False)
fig.show()
```
