```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

ipc_desempleo = pd.read_excel('datos_connected_scattered_chart.xlsx').iloc[:12]
ipc_desempleo

df = ipc_desempleo.copy()
df['Fecha'] = pd.to_datetime(df['Fecha'])
df = df.sort_values('Fecha').reset_index(drop=True)
df['MesAño'] = df['Fecha'].dt.strftime('%b %Y')

x = df['IPC']
y = df['Nóminas no agrícolas']

pos = np.where(np.arange(len(df)) % 2 == 0, 'top center', 'bottom center')

THEME_COLOR = '#3b82f6'
EDGE_COLOR  = 'rgba(30,41,59,0.25)'
TEXT_COLOR  = '#1f2937'
GRID_COLOR  = 'rgba(0,0,0,0.05)'

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=x, y=y, mode='lines',
    line=dict(width=2, color=THEME_COLOR),
    hoverinfo='skip', showlegend=False
))

fig.add_trace(go.Scatter(
    x=x, y=y,
    mode='markers+text',
    text=df['MesAño'],
    textposition=pos,
    textfont=dict(size=12, color=TEXT_COLOR),
    marker=dict(size=9, color=THEME_COLOR, line=dict(width=1, color=EDGE_COLOR)),
    showlegend=False,
    hovertemplate="<b>%{text}</b><br>IPC: %{x:.1f}%<br>Nóminas: %{y:.0f}<extra></extra>"
))

fig.add_annotation(x=x.iloc[0], y=y.iloc[0], text="Inicio",
                   showarrow=True, arrowhead=3, ax=-28, ay=26,
                   font=dict(size=11, color=TEXT_COLOR),
                   bgcolor="rgba(255,255,255,0.8)")
fig.add_annotation(x=x.iloc[-1], y=y.iloc[-1], text="Fin",
                   showarrow=True, arrowhead=3, ax=28, ay=-24,
                   font=dict(size=11, color=TEXT_COLOR),
                   bgcolor="rgba(255,255,255,0.8)")

fig.update_layout(
    template='plotly_white',
    title='Connected Scatter: IPC vs. Nóminas no agrícolas de EE. UU.',
    title_x=0.02,
    font=dict(size=12, color=TEXT_COLOR),
    margin=dict(l=56, r=28, t=60, b=48)
)

fig.update_xaxes(
    title_text='IPC',
    tickformat=".1f",
    ticksuffix="%",
    showline=True, linewidth=0.9, linecolor='rgba(0,0,0,0.65)',
    ticks='outside', ticklen=4,
    zeroline=False,
    showgrid=True, gridcolor=GRID_COLOR
)

fig.update_yaxes(
    title_text='Nóminas no agrícolas (miles de personas)',
    showline=True, linewidth=0.9, linecolor='rgba(0,0,0,0.65)',
    ticks='outside', ticklen=4,
    zeroline=False,
    showgrid=True, gridcolor=GRID_COLOR
)

fig.show()
```
